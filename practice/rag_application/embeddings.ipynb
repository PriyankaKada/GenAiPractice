```python
import ollama

CHAT_MODEL = "llama3.1:8b"

client = ollama.Client(host="http://localhost:11434")

response = client.chat(
    model=CHAT_MODEL,
    messages=[
        {"role": "system", "content": "You are a consise assistant."},
        {"role": "user", "content": "why do people vote donald trump?"}
    ],
)
print(response)
```

```python
import ollama
import numpy as np

EMBEDDING_MODEL = "embeddinggemma:latest"
client = ollama.Client(host="http://localhost:11434")

response = client.embed(
    model=EMBEDDING_MODEL,  
    input=["quick lazy fox jumped over lazy dog", "the quick brown fox jumped over the lazy dog"]
)

# 1. Print the model name (accessed via dictionary key or dot notation object)
print(f"Model Used: {EMBEDDING_MODEL}")

# 2. Extract vectors from the plural 'embeddings' list
vector1 = np.array(response["embeddings"][0])
vector2 = np.array(response["embeddings"][1])

# 3. Print shapes to confirm dimensionality
print(f"Vector 1 Shape: {vector1.shape}")
print(f"Vector 2 Shape: {vector2.shape}")

# 4. Calculate cosine similarity
```
